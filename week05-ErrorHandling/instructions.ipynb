### Password System Instructions

You have been provided with a password interface that people can use to login to some account. The graphical user interface (GUI) has already been built for you, but it is your task to implement the backend of the password system where the usernames and passwords are checked for validity. The challenge is that you have to implement the backend **WITHOUT** modifying `main.py` or `gui.py` **AT ALL**. Start by reading through the existing code, in `gui.py` the main thing you want to pay attention to is the `submit_credentials` function.  

Note: This is purely for practicing custom errors and exception handling, this is not how you should go about building an actual password system. Please **never** actually send unencrypted login credentials from a frontend to a server!

#### Task 0: Making data

- Create a text file `user_data.txt` that will be used to store usernames and passwords.

- Create several username/password combinations that you can use to test your login system.

- *Note: If you are ever designing a real login system. Please never store passwords in a plaintext file. I hope the reason for that is obvious.*


#### Task 1: Password validation

- We have the user interface already created, but we need to implement the system that actually checks to see if passwords are correct. In `backend.py` implement the `verify_credentials` function:

    - If either the username or password submission is empty, raise a custom `MissingCredentialError` and have the error message specify whether it was the username or password that was empty. 

    - Check the entered username and password against the passwords you created in **Task 0**. If the username and password don't match, raise a custom `InvalidCredentialsError`.
        - Look at the `submit_credentials` function in `gui.py`, what does your `verify_credentials` function need to do if the username and password are correct?

#### Task 2: Too many login attempts

- If someone tries to login multiple times with the same username and incorrect password, they need to be prevented from attempting to login again for a period of time to avoid people just trying to brute force guess someone's password. Implement the following features:

    - For all usernames, create a system that tracks how many times someone has tried to login with an incorrect password since the last successful login.
        - Ex. If Sierra messes up her password twice, she has 2 failed login attempts. If she successfully logs in on the next attempt, the number gets reset to 0.

    - If someone enters the wrong password **more** than 3 times in a row, raise a custom `MaximumTriesExceededError`. This error should take as a parameter, the time when the error was raised so that you can determine when they should regain access.
        - A `MissingCredentialError` should not count as incorrectly entering the password.
        - Hint: Using the `time` library, you can access the current time in seconds using `time.time()`.

    - The next time someone tries to login with the **same username**, check if it has been at least 30 seconds since the last login attempt.
        - If yes, process the login attempt as usual.
        - If no, raise a `MaximumTriesExceededError` regardless of whether the password was correct.

- Consider storing the number of failed login attempts in an external file, this will make sure the data is maintained even if your "server" (python code) stops running.
